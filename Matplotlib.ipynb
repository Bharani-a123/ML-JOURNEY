```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

 simple Graph with basic Using Line


```python
x=np.array([2,3,4,5,6,7])
y=np.sin(x)
plt.plot(x,y)
plt.show()
```

```python
#with labels
x=np.array([2,3,4,5,6,7])
y=np.sin(x)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Simple Grap")
plt.plot(x,y)
plt.show()
```

'o'  → circle  
's'  → square  
'^'  → triangle up  
'v'  → triangle down  
'*'  → star  
'+'  → plus  
'x'  → x mark  
'D'  → diamond  
'p'  → pentagon  
'h'  → hexagon



Line styles (linestyle= or ls=)
Controls how the line is drawn.

'-'   → solid line  
'--'  → dashed line  
'-.'  → dash-dot line  
':'   → dotted line  
'None' or '' → no line (only markers)

```python
x=np.array([2,3,4,5,6,7])
y=np.sin(x)
plt.figure(figsize=(2,2))#makes size of canva to plot graph
plt.grid(True)#provide grid like structure


plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Simple Grap")
plt.plot(x,y,marker='D',linestyle=':')
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(8, 6))
plt.plot(x, y, marker='o', linestyle='-')

# Add annotations
for i, (xi, yi) in enumerate(zip(x, y)):
    plt.annotate(f'({xi}, {yi})', (xi, yi), textcoords="offset points", xytext=(0, 10), ha='center')

plt.title('Line Chart with Annotations')
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4])
y = x*2

plt.plot(x, y)

x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]

plt.plot(x, y1, '-.')
plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title('multiple plots')

plt.fill_between(x,y,y1, color='green', alpha=0.5)
plt.show()
```

# **BAR GRAPH **

```python
import matplotlib.pyplot as plt
import numpy as np

fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]

plt.bar(fruits, sales,color='red')
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

```python
fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]

plt.barh(fruits, sales,color='red')#makes horizontal plot
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```
